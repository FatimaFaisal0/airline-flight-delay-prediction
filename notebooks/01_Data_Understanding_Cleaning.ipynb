## Part A – Data Understanding

```python
# Import the required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Display more columns in Jupyter
pd.set_option('display.max_columns', None)
```

```python
# Load the original airline dataset
df = pd.read_csv("T_ONTIME_REPORTING.csv", low_memory=False)

# Display the first five rows
df.head()
```

```python
# Check the number of rows and columns
print("Dataset Shape:", df.shape)

# Display total rows
print("Total Rows:", df.shape[0])

# Display total columns
print("Total Columns:", df.shape[1])

# Display column names
print("\nColumn Names:")
print(df.columns.tolist())
```

```python
# Display information about the dataset
df.info()

# Check the data type of each column
print("\nData Types:")
print(df.dtypes)
```

```python
# Display statistical summary of numerical columns
print("Numerical Summary:")
display(df.describe())

# Display summary of all columns
print("Summary of All Columns:")
display(df.describe(include="all"))
```

## Part B – Data Quality Checks

```python
# Count missing values in each column
missing_values = df.isnull().sum()

# Calculate percentage of missing values
missing_percentage = (missing_values / len(df)) * 100

# Create a missing values summary table
missing_summary = pd.DataFrame({
    "Missing Values": missing_values,
    "Missing Percentage": missing_percentage.round(2)
})

# Display columns with missing values
display(missing_summary[
    missing_summary["Missing Values"] > 0
].sort_values("Missing Values", ascending=False))
```

```python
# Select columns containing missing values
missing_plot = missing_summary[
    missing_summary["Missing Values"] > 0
]

# Create bar chart
plt.figure(figsize=(10, 5))
plt.bar(missing_plot.index, missing_plot["Missing Values"])

# Add chart labels
plt.title("Missing Values in Airline Dataset")
plt.xlabel("Columns")
plt.ylabel("Number of Missing Values")

# Improve readability
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
# Check the number of exact duplicate rows
duplicate_count = df.duplicated().sum()

print("Number of Duplicate Rows:", duplicate_count)

# Display duplicate records if any exist
display(df[df.duplicated(keep=False)].head(10))
```

```python
# Check unique values for cancellation status
print("Cancellation Values:")
print(df["CANCELLED"].value_counts(dropna=False))

# Check unique values for diversion status
print("\nDiversion Values:")
print(df["DIVERTED"].value_counts(dropna=False))

# Check unique target values
print("\nArrival Delay Classification:")
print(df["ARR_DEL15"].value_counts(dropna=False))

# Check invalid scheduled flight durations
invalid_duration = df[df["CRS_ELAPSED_TIME"] <= 0]

print("\nInvalid Scheduled Durations:", len(invalid_duration))

# Display unusual records
display(invalid_duration[
    ["FL_DATE", "ORIGIN", "DEST", "CRS_ELAPSED_TIME"]
].head(10))
```

```python
# Convert flight date into datetime format
date_check = pd.to_datetime(
    df["FL_DATE"],
    format="%m/%d/%Y %H:%M",
    errors="coerce"
)

# Check invalid dates
print("Invalid Flight Dates:", date_check.isna().sum())

# Check minimum and maximum dates
print("Earliest Flight Date:", date_check.min())
print("Latest Flight Date:", date_check.max())

# Check scheduled time validity
def invalid_time(series):
    time = pd.to_numeric(series, errors="coerce")
    hours = time // 100
    minutes = time % 100

    valid = (
        (hours.between(0, 23) & minutes.between(0, 59))
        | (time == 2400)
    )

    return (~valid | time.isna()).sum()

print("Invalid Scheduled Departure Times:",
      invalid_time(df["CRS_DEP_TIME"]))

print("Invalid Scheduled Arrival Times:",
      invalid_time(df["CRS_ARR_TIME"]))
```

```python
# Create a copy of the original dataset
df_clean = df.copy()

# Keep the original number of records
original_rows = len(df_clean)

print("Original Rows:", original_rows)
```

```python
# Convert the flight date to datetime
df_clean["FL_DATE"] = pd.to_datetime(
    df_clean["FL_DATE"],
    format="%m/%d/%Y %H:%M",
    errors="coerce"
)

# Select categorical columns
text_columns = [
    "OP_UNIQUE_CARRIER",
    "ORIGIN",
    "ORIGIN_CITY_NAME",
    "ORIGIN_STATE_ABR",
    "DEST",
    "DEST_CITY_NAME",
    "DEST_STATE_ABR",
    "DEP_TIME_BLK",
    "ARR_TIME_BLK"
]

# Remove unnecessary spaces from text values
for col in text_columns:
    df_clean[col] = df_clean[col].astype("string").str.strip()

# Standardize airport and airline codes
code_columns = [
    "OP_UNIQUE_CARRIER",
    "ORIGIN",
    "ORIGIN_STATE_ABR",
    "DEST",
    "DEST_STATE_ABR"
]

for col in code_columns:
    df_clean[col] = df_clean[col].str.upper()

# Display updated data types
print("Updated Data Types:")
print(df_clean.dtypes)
```

```python
# Check missing cancellation codes by cancellation status
print("Missing Cancellation Codes by Status:")
print(pd.crosstab(
    df_clean["CANCELLED"],
    df_clean["CANCELLATION_CODE"].isna()
))

# Fill missing cancellation codes for non-cancelled flights
df_clean.loc[
    (df_clean["CANCELLED"] == 0) &
    (df_clean["CANCELLATION_CODE"].isna()),
    "CANCELLATION_CODE"
] = "Not Cancelled"

# Remove exact duplicate records, if any
df_clean = df_clean.drop_duplicates().copy()

print("Rows After Removing Duplicates:", len(df_clean))
print("Duplicates Removed:", original_rows - len(df_clean))
```

```python
# Create a function to identify valid scheduled times
def valid_time(series):
    time = pd.to_numeric(series, errors="coerce")
    hours = time // 100
    minutes = time % 100

    return (
        (hours.between(0, 23) & minutes.between(0, 59))
        | (time == 2400)
    )

# Identify records with invalid scheduled information
invalid_records = (
    df_clean["FL_DATE"].isna()
    | (df_clean["CRS_ELAPSED_TIME"] <= 0)
    | df_clean["CRS_ELAPSED_TIME"].isna()
    | (df_clean["DISTANCE"] <= 0)
    | df_clean["DISTANCE"].isna()
    | ~valid_time(df_clean["CRS_DEP_TIME"])
    | ~valid_time(df_clean["CRS_ARR_TIME"])
)

print("Invalid Records:", invalid_records.sum())

# Save invalid records for documentation
invalid_flights = df_clean[invalid_records].copy()

display(invalid_flights.head(10))
```

```python
# Remove records with invalid essential flight information
df_clean = df_clean.loc[~invalid_records].copy()

# Check remaining records
print("Rows After Cleaning Invalid Records:", len(df_clean))

# Check for remaining invalid scheduled durations
print("Remaining Invalid Durations:",
      (df_clean["CRS_ELAPSED_TIME"] <= 0).sum())
```

```python
# Select columns to investigate for outliers
outlier_columns = [
    "DISTANCE",
    "CRS_ELAPSED_TIME",
    "DEP_DELAY",
    "ARR_DELAY"
]

# Create a list to store the results
outlier_results = []

# Calculate outliers using the IQR method
for col in outlier_columns:

    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outlier_count = (
        (df_clean[col] < lower_bound) |
        (df_clean[col] > upper_bound)
    ).sum()

    outlier_results.append({
        "Column": col,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Outlier Count": outlier_count
    })

# Create outlier summary table
outlier_summary = pd.DataFrame(outlier_results)

display(outlier_summary)
```

```python
# Create boxplots for numerical variables
for col in outlier_columns:

    plt.figure(figsize=(9, 3))

    plt.boxplot(df_clean[col].dropna(), vert=False)

    plt.title("Outlier Analysis: " + col)
    plt.xlabel(col)

    plt.tight_layout()
    plt.show()
```

## Part D – Prepare the Dataset for Classification

```python
# Count flights with missing arrival delay classification
missing_target = df_clean["ARR_DEL15"].isna()

print("Missing Target Values:", missing_target.sum())

# Investigate cancelled and diverted flights
print("\nCancelled and Diverted Flights:")
print(pd.crosstab(
    [df_clean["CANCELLED"], df_clean["DIVERTED"]],
    missing_target
))
```

```python
# Keep flights that were not cancelled or diverted
df_model = df_clean[
    (df_clean["CANCELLED"] == 0) &
    (df_clean["DIVERTED"] == 0)
].copy()

# Remove records with missing target labels
df_model = df_model.dropna(subset=["ARR_DEL15"])

# Keep only valid classification labels
df_model = df_model[
    df_model["ARR_DEL15"].isin([0, 1])
].copy()

# Convert target to integer
df_model["ARR_DEL15"] = df_model["ARR_DEL15"].astype(int)

print("Classification Dataset Shape:", df_model.shape)

print("\nTarget Distribution:")
print(df_model["ARR_DEL15"].value_counts())

print("\nMissing Target Values:",
      df_model["ARR_DEL15"].isna().sum())
```

```python
# Select features available before departure
feature_columns = [
    "MONTH",
    "DAY_OF_MONTH",
    "DAY_OF_WEEK",
    "OP_UNIQUE_CARRIER",
    "ORIGIN",
    "DEST",
    "CRS_DEP_TIME",
    "CRS_ARR_TIME",
    "CRS_ELAPSED_TIME",
    "DISTANCE"
]

# Select model input features
X = df_model[feature_columns].copy()

# Select target variable
y = df_model["ARR_DEL15"].copy()

# Check input and target shapes
print("Input Features Shape:", X.shape)
print("Target Shape:", y.shape)

# Check missing values in selected features
print("\nMissing Input Values:")
print(X.isnull().sum())

# Display selected input features
display(X.head())
```

## Part E – Final Checks and Save Results

```python
# Create a summary of data cleaning results
cleaning_summary = pd.DataFrame({
    "Stage": [
        "Original Dataset",
        "After Data Cleaning",
        "Classification Dataset"
    ],
    "Rows": [
        len(df),
        len(df_clean),
        len(df_model)
    ],
    "Columns": [
        df.shape[1],
        df_clean.shape[1],
        df_model.shape[1]
    ]
})

# Display the cleaning summary
display(cleaning_summary)

# Check exact duplicates after cleaning
print("Remaining Duplicates:", df_clean.duplicated().sum())

# Compare missing values before and after cleaning
comparison = pd.DataFrame({
    "Before Cleaning": df.isna().sum(),
    "After Cleaning": df_clean.isna().sum()
})

display(comparison)
```

```python
# Save cleaned dataset
df_clean.to_csv(
    "Airline_Cleaned_Data.csv",
    index=False
)

# Save classification-ready dataset
df_model.to_csv(
    "Airline_Classification_Data.csv",
    index=False
)

# Save selected model features and target
model_data = X.copy()
model_data["ARR_DEL15"] = y

model_data.to_csv(
    "Airline_Model_Features.csv",
    index=False
)

# Save missing-value summary
missing_summary.to_csv(
    "Missing_Values_Summary.csv"
)

# Save outlier summary
outlier_summary.to_csv(
    "Outlier_Summary.csv",
    index=False
)

# Save cleaning summary
cleaning_summary.to_csv(
    "Cleaning_Summary.csv",
    index=False
)

# Save invalid records for documentation
invalid_flights.to_csv(
    "Invalid_Flight_Records.csv",
    index=False
)

print("All Stage 3 files saved successfully!")
```
